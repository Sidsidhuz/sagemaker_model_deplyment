```python
from sagemaker.pytorch import PyTorchModel
import sagemaker

role = sagemaker.get_execution_role()

model = PyTorchModel(
    model_data="s3://bucket-sid-aws/model.tar.gz",  # model from S3
    role=role,
    entry_point="inference.py",
    source_dir="code",
    framework_version="1.13",
    py_version="py39"
)

predictor = model.deploy(
    instance_type="ml.t2.medium",
    initial_instance_count=1
)

print("Successfully deployed")
```

```python
from sagemaker.serializers import JSONSerializer
from sagemaker.deserializers import JSONDeserializer

# This forces SageMaker to stop trying to decode the 0x93 byte as text
predictor.serializer = JSONSerializer()
predictor.deserializer = JSONDeserializer()

import json

with open("test1.jpg", "rb") as f:
    img_hex = f.read().hex()

data = {
    "crop": "Grapes",
    "image": img_hex
}

# Use the predictor directly with the dictionary
# The JSONSerializer will handle encoding correctly

response = {"prediction": predictor.predict(data), "crop": data["crop"]}
```
